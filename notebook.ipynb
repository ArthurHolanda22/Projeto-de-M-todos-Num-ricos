```python
tol = 1e-6

# Execução dos métodos
h_bis, iter_bis, err_bis = mn.bissecao(mn.f, a=0.1, b=10.0, tol=tol)
h_new, iter_new, err_new = mn.newton_raphson(mn.f, mn.df, x0=2.0, tol=tol)
h_sec, iter_sec, err_sec = mn.secante(mn.f, x0=1.0, x1=2.0, tol=tol)

print(f"{'Método':<18} | {'Altura H (m)':<15} | {'Iterações':<10}")
print("-" * 50)
print(f"{'Bisseção':<18} | {h_bis:<15.6f} | {iter_bis:<10}")
print(f"{'Newton-Raphson':<18} | {h_new:<15.6f} | {iter_new:<10}")
print(f"{'Secante':<18} | {h_sec:<15.6f} | {iter_sec:<10}")
```

```python
H_vals = np.linspace(0.1, 5.0, 200)
f_vals = [mn.f(h) for h in H_vals]

plt.figure(figsize=(8, 4))
plt.plot(H_vals, f_vals, label=r'$f(H) = v(H) - 5$', color='navy')
plt.axhline(0, color='red', linestyle='--', label='f(H) = 0')
plt.axvline(h_new, color='green', linestyle=':', label=f'Solução H ≈ {h_new:.4f} m')

plt.title('Determinação Numérica da Altura H')
plt.xlabel('Altura H (m)')
plt.ylabel('f(H)')
plt.legend()
plt.grid(True)
plt.show()
```
